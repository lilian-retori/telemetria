# EDA — Telemetria de ônibus

Este notebook é opcional. Use-o para explorar os dados sem duplicar a lógica do pipeline. A lógica de produção deve ficar em `src/telemetria/`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

from telemetria.config import Config
from telemetria.io import carregar_telemetria
from telemetria.validation import padronizar_colunas, validar_registros
from telemetria.features import adicionar_metricas_movimento
from telemetria.metrics import resumir_por_hora
```

```python
config = Config()
brutos = carregar_telemetria(config.input_path)
padronizados = padronizar_colunas(brutos)
aceitos, rejeitados = validar_registros(padronizados, config)
telemetria = adicionar_metricas_movimento(aceitos, config)
resumo_horario = resumir_por_hora(telemetria)
resumo_horario.head()
```

```python
ax = resumo_horario.plot(x='HORA', y='percentual_baixa_velocidade', figsize=(12, 4), legend=False)
ax.set_ylabel('Baixa velocidade (%)')
ax.set_xlabel('Hora')
ax.set_title('Percentual de baixa velocidade por hora')
plt.show()
```
